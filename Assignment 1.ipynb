### Conceptual Background: Understanding the Training Process

To strictly solve the task, we need to understand the four key concepts that drive the training loop:

*   **Forward Pass**: We calculate the model's prediction based on current inputs and weights.
*   **Loss**: We quantify how "wrong" the model is by comparing the prediction to the actual target.
*   **Gradient**: We find the slope of the loss function with respect to the weights. This tells us which direction (increase/decrease) and how much to shift the weights to reduce the error.
*   **Backpropagation**: We essentially use the gradient to update our weights in the opposite direction of the error, effectively "learning" from the data.

### Theoretical Explanation of the Formulas

Here is the "why" behind the mathematical steps shown in the task description:

1.  **Forward Pass**: $\hat{y} = x \cdot w$
    *   *Theory*: This represents a simple linear neuron without a bias term. It predicts the output $\hat{y}$ by scaling the input $x$ by parameter weight $w$.

2.  **Loss Calculation**: $L = (t - \hat{y})^2$
    *   *Theory*: We use the Squared Error loss for this sample. It penalizes predictions that are far from the true target $t$. Squaring ensures the error is always positive and penalizes larger errors more heavily than smaller ones.

3.  **Gradient Calculation**: $\frac{\partial L}{\partial w} = 2x(\hat{y} - t)$
    *   *Theory*: This is the derivative of the loss with respect to the weight, derived using the chain rule: $\frac{\partial L}{\partial w} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial w}$. It indicates the direction and rate at which the Loss changes as we change $w$.
    *   Inner derivatives: $\frac{\partial L}{\partial \hat{y}} = 2(\hat{y} - t)$ and $\frac{\partial \hat{y}}{\partial w} = x$.

4.  **Parameter Update**: $w_{new} = w_{old} - \alpha \cdot \frac{\partial L}{\partial w}$
    *   *Theory*: This is the Gradient Descent update rule. We subtract a fraction ($\alpha$, the learning rate) of the gradient from the old weight. This moves the weight "downhill" along the error surface towards a minimum loss.

### Solution Code To Task A

```python
import pandas as pd

# 1. Dataset (First 3 samples)
# X = Input, T = Target (Y in description)
data = [
    {'sample': 1, 'x': 0.870968, 't': 0.929654},
    {'sample': 2, 'x': 0.153226, 't': 0.109427},
    {'sample': 3, 'x': 0.104839, 't': 0.037186}
]

# 2. Hyperparameters (Student Chosen)
# Please change these values to your own unique choice!
learning_rate = 0.1   # Alpha (Learning Rate)
w_current = 0.5       # Initial Weight (w_old)

print(f"Initial Weight (w_old): {w_current}")
print(f"Learning Rate (alpha): {learning_rate}")
```

```python
# Sample 1
w_old = w_current
x1 = data[0]['x']
t1 = data[0]['t']

print(f"--- Sample 1 (x={x1:.6f}, t={t1:.6f}) ---")
y_hat1 = x1 * w_old
loss1 = (t1 - y_hat1) ** 2
print("Sample 1 Loss:", loss1)

gradient1 = 2 * x1 * (y_hat1 - t1)
print("Sample 1 Gradient:", gradient1)

w_new = w_old - (learning_rate * gradient1)
print("Sample 1 New w:", w_new)


# Sample 2 (IMPORTANT: start from the new weight from Sample 1)
w_old = w_new
x2 = data[1]['x']
t2 = data[1]['t']

print(f"\n--- Sample 2 (x={x2:.6f}, t={t2:.6f}) ---")
y_hat2 = x2 * w_old
loss2 = (t2 - y_hat2) ** 2
print("Sample 2 Loss:", loss2)

gradient2 = 2 * x2 * (y_hat2 - t2)
print("Sample 2 Gradient:", gradient2)

w_new = w_old - (learning_rate * gradient2)
print("Sample 2 New w:", w_new)


# Sample 3 (IMPORTANT: start from the new weight from Sample 2)
w_old = w_new
x3 = data[2]['x']
t3 = data[2]['t']

print(f"\n--- Sample 3 (x={x3:.6f}, t={t3:.6f}) ---")
y_hat3 = x3 * w_old
loss3 = (t3 - y_hat3) ** 2
print("Sample 3 Loss:", loss3)

gradient3 = 2 * x3 * (y_hat3 - t3)
print("Sample 3 Gradient:", gradient3)

w_new = w_old - (learning_rate * gradient3)
print("Sample 3 New w:", w_new)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Reconstruct the Weight History for Plotting ---
# (We re-calculate the steps programmatically to ensure the plot matches the manual findings)
w_history = [w_current]
w_temp = w_current

for item in data:
    x = item['x']
    t = item['t']
    # SGD Update Step
    prediction = x * w_temp
    gradient = 2 * x * (prediction - t)
    w_temp = w_temp - (learning_rate * gradient)
    w_history.append(w_temp)

# --- 2. Calculate Global Loss Curve ---
# Define a range of weights to visualize the "error landscape"
ws = np.linspace(0, 1.2, 100) 

# Function to calculate Mean Squared Error (MSE) for the whole dataset (3 samples)
def batch_mse(w, data):
    total_error = 0
    for item in data:
        total_error += (item['t'] - item['x'] * w) ** 2
    return total_error / len(data)

# Calculate loss for every weight in our range
loss_ws = [batch_mse(w, data) for w in ws]
# Calculate loss for our specific SGD steps
history_loss = [batch_mse(w, data) for w in w_history]

# --- 3. Plotting ---
plt.figure(figsize=(10, 6))

# Plot the theoretical Loss Curve (Red Dashed Line)
plt.plot(ws, loss_ws, 'r--', label='Global Loss Function (MSE)', linewidth=2)

# Plot the User's SGD Path (Blue Line & Dots)
plt.scatter(w_history, history_loss, color='blue', s=100, label='SGD Steps', zorder=5)
plt.plot(w_history, history_loss, color='blue', linestyle='-', linewidth=1.5, label='Weight Trajectory')

# Annotation for clarity
for i, w in enumerate(w_history):
    label = "Init" if i == 0 else f"Step {i}"
    plt.annotate(label, (w, history_loss[i]), textcoords="offset points", xytext=(0,10), ha='center', fontsize=9, fontweight='bold')

plt.title('Figure 1: SGD Optimization Path: Movement of $w$ on Loss Function Surface')
plt.xlabel('Weight ($w$)')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Result Analysis and Conclusion

After running the manual SGD steps above, we can observe clear evidence of "learning":

1.  **Loss Reduction (Error Improvement)**:
    *   **Sample 1**: The loss starts relatively high compared to the next few samples (around **0.244**). This is expected as the initial weight ($w=0.5$) was just a guess.
    *   **Sample 2 & 3**: The loss drops drastically (to approx **0.000385** and **0.000591**). This large improvement show that the training process from Sample 1 moved the weight significantly closer to the correct relationship.

2.  **Weight Optimization**:
    *   The weight shifted from **0.5** to approx **0.586** in the first step.
    *   In subsequent steps, the weight changes are much smaller (stabilizing around **0.5687 (S2)** and **0.5682(S3)**). This indicates that the model has quickly found a weight that works well for this data.

### Why Does The Loss Values in The Plot Differ From Manual Calculations?

you might notice that the manual calculation for **Sample. 1** gave a loss of **0.244**, but the first point in figure 1 is much lower (around **0.08**). This is expected!

The answer lies in the **difference between Batch Gradient Descent and Stochastic Gradient Descent (SGD)**:

1.  **Batch Gradient Descent**.
    *   With BGD, loss is calculated using **all data points at once**: $Loss = MSE_{global}$.
    *   Therefore, a manual loss value determined from BGD would be the same as the value on the Global Loss curve (**Red Dashed Curve**).
    
2.  **Stochastic Gradient Descent**.
    *   With SGD, I calculate loss using **only ONE data point**: $Loss = (t - \hat{y}_{sample})^2$. This is called "Stochastic Loss".
    *   **Cell 5 Prints**: The Stochastic Loss (e.g., $0.244$ for Sample 1).
    *   **Cell 6 Plots**: The Global Loss / MSE (Average of all 3 sample loss calculations determined from the underlying selected Weight). 
    *   **For Sample 1** $w=0.5$, therefore $\frac{0.244203 + 0.000385 + 0.000591}{3} \approx 0.0817$.

**Visualizing Effect**:
By plotting the *weight's position* on the *Global Loss Curve*, we can see if the noisy SGD updates are actually moving us towards the global minimum (the bottom of the valley), which is the true goal of training.

### A Few Considerations Regarding The Red Curve (Global Loss)

**The Trade-off (and how the curve is determined):**
The red curve represents the **"Error Landscape"**. To construct it, we **programmatically sweep through a range of 100 theoretical weights** (e.g. from 0 to 1.2). For **each** of these 100 weights, we calculate the Mean Squared Error for the **full dataset** (all samples simultaneously). This creates a "ground truth" map that visually demonstrates the trade-off: there is a "sweet spot" (minimum) where the weight is neither too high nor too low.

**Why not just "solve" the curve instead of machine learning?**
In this simple example with 1 parameter, we *could* easily calculate the minimum mathematically (using calculus) or just look at the plot to find the answer. However, in real **Deep Learning**, models have **millions or billions of parameters**.
*   **The Curse of Dimensionality**: It is computationally impossible to "plot" or "see" the landscape across such a huge range of combinations for millions of dimensions. This is why we need iterative Machine Learning algorithms like SGD instead of just "solving" it.

### Effect on training process if we change the hyperparameter and initial weight

```python
# Define different scenarios to test
scenarios = [
    {'name': 'Original (LR=0.1, w=0.5)', 'lr': 0.1, 'w_init': 0.5, 'color': 'blue'},
    {'name': 'Low LR (LR=0.01, w=0.5)', 'lr': 0.01, 'w_init': 0.5, 'color': 'green'},
    {'name': 'High LR (LR=1.1, w=0.5)', 'lr': 1.1, 'w_init': 0.5, 'color': 'red'}, # Deliberately high to show overshooting
    {'name': 'Diff Init (LR=0.1, w=0.0)', 'lr': 0.1, 'w_init': 0.0, 'color': 'orange'}
]

plt.figure(figsize=(12, 7))

# 1. Plot the Background Global Loss Curve (Reference)
ws = np.linspace(-0.2, 1.4, 100) 
loss_ws = [batch_mse(w, data) for w in ws]
plt.plot(ws, loss_ws, 'k--', label='Global Loss Function', alpha=0.3)

# 2. Run and Plot Each Scenario
for s in scenarios:
    w_curr = s['w_init']
    w_path = [w_curr]
    
    # Run the same SGD loop for this scenario
    for item in data:
        x_val = item['x']
        t_val = item['t']
        prediction = x_val * w_curr
        gradient = 2 * x_val * (prediction - t_val)
        w_curr = w_curr - (s['lr'] * gradient)
        w_path.append(w_curr)
    
    # Calculate loss for the path
    loss_path = [batch_mse(w, data) for w in w_path]
    
    # Plotting
    plt.plot(w_path, loss_path, marker='o', label=s['name'], color=s['color'], linewidth=2, alpha=0.8)
    
    # Mark the start and end of each path
    plt.text(w_path[0], loss_path[0], 'Start', color=s['color'], fontsize=8, ha='right')
    plt.text(w_path[-1], loss_path[-1], 'End', color=s['color'], fontsize=8, ha='left')

plt.title('Figure 2: Effect of Hyperparameters (Learning Rate) & Initial Weight on Training')
plt.xlabel('Weight ($w$)')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Interpretation of Results in Figure 2

1.  **Low Learning Rate (Green)**: The updates are tiny. The model learns very slowly and doesn't reach the bottom of the valley (optimal weight) within just 3 steps. It requires many more epochs to converge.
2.  **High Learning Rate (Red)**: The updates are huge. The weight jumps drastically, potentially shooting past the minimum (overshooting) or even diverging (moving away from the solution). Notice how it might jump to the other side of the curve!
3.  **Different Initialization (Orange)**: Starting at $w=0.0$ means we start further away from the optimal weight (approx 1.0). However, with a good learning rate, it still moves in the correct direction efficiently.
4.  **Conclusion**: The choice of Learning Rate is a trade-off. Too low = slow training. Too high = unstable training. Finding the 'Goldilocks' zone (like our blue line) is key to deep learning. Also, the choice of initial weight can have a significant effect on the training process time and computational steps

# Assignment 1 - Task B: Attention Contextualization

![Assignment 1 Task B](Assignment%201%20task%20B.png)

### Conceptual Background: The Problem with Static Embeddings
Standard word embeddings (like the ones we will initialize below) are **static**. This means a word like "bat" has the exact same vector representation $\vec{v}_{bat}$ regardless of whether it appears in "The bat flew over his head" (animal) or "He hit the ball with a bat" (equipment). This is a limitation when dealing with **polysemy** (words with multiple meanings).

**The Task**:
We select the homonym **"Bat"**.
*   **Sentence 1**: "The bat flew over his head" (Animal Context)
*   **Sentence 2**: "He hit the ball with a bat" (Sports Context)

We will use **Self-Attention** to calculate new, **contextualized** vectors. The mechanism allows the word "bat" to "pay attention" to relevant context words (like "flew" or "hit"), effectively shifting its representation towards the correct meaning.

### Theoretical Explanation: Self-Attention Mechanism
The Self-Attention mechanism updates a token's representation by taking a weighted sum of all tokens in the sequence (including itself). The process involves:

1.  **Query, Key, and Value ($Q, K, V$)**: In this simplified example (and typically in the first layer), we assume $Q = K = V = E$ (the input embeddings).
2.  **Attention Scores (Similarity)**: Computed by the dot product $Q \cdot K^T$. This tells us how relevant each word is to every other word. High dot product = high similarity/relevance.
3.  **Attention Weights (Softmax)**: We apply the softmax function to normalize scores into probabilities (summing to 1). $\text{softmax}(x_i) = \frac{e^{x_i}}{\sum e^{x_j}}$.
4.  **Contextualized Output ($A$)**: The final vector is the weighted sum of Value vectors: $A = \text{weights} \cdot V$.

Attention Formula from the article "Attention Is All You Need":
$$ \text{Attention}(Q, K, V)
= \text{softmax}\!\left(\frac{QK^{T}}{\sqrt{d_k}}\right) V $$
*(Note: In full Transformer papers, we often divide by $\sqrt{d_k}$ to scale the dot products, but we will follow the simplified approach from the lecture notebook, "M3_2_Transformermodels_NLU_v3.ipynb").*
$$ \text{Attention}(Q, K, V) = \text{softmax}(Q K^T) V $$

### Solution Code to Task B

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Define 2D Embeddings for a small vocabulary
# We use the homonym "Bat" (Animal vs Sports)
embeddings = {
    "the": np.array([0.1, 0.1]),
    "bat": np.array([0.5, 0.5]),  # The homonym (Centered/Ambiguous)
    "flew": np.array([0.1, 0.9]), # Animal context (High Y)
    "over": np.array([0.15, 0.8]), # Animal context
    "his": np.array([0.1, 0.2]),
    "head": np.array([0.2, 0.85]), # Animal/Body context
    "he": np.array([0.7, 0.2]), 
    "hit": np.array([0.9, 0.1]),  # Sports context (High X)
    "ball": np.array([0.85, 0.1]), # Sports context
    "with": np.array([0.05, 0.05]),
    "a": np.array([0.1, 0.15])
}

# 2. Define Sentences
# Lowercase to match keys in embeddings
sentence_1 = ["the", "bat", "flew", "over", "his", "head"]
sentence_2 = ["he", "hit", "the", "ball", "with", "a", "bat"]

# 3. Helper Function: Softmax
def softmax(x):
    # Subtract max for numerical stability
    e_x = np.exp(x - np.max(x))
    # Sum over the last axis (columns) to normalize each row (query)
    # This ensures each word distributes its attention to others summing to 1
    return e_x / e_x.sum(axis=1, keepdims=True)

# 4. Helper Function: Self-Attention Calculation
def calculate_self_attention(sentence, embedding_dict):
    # Stack embeddings to form the matrix (Shape: Sequence_Length x Embedding_Dim)
    # For Self-Attention: Q = K = V = Inputs
    X = np.array([embedding_dict[word] for word in sentence])
    Q = K = V = X
    
    # Step 1: Similarity (Dot Product)
    # Q dot K^T
    scores = Q.dot(K.T)
    
    # Step 2: Weights (Softmax)
    weights = softmax(scores)
    
    # Step 3: Contextualized Embeddings (Weighted Sum)
    # Weights dot V
    context_embeddings = weights.dot(V)
    
    return X, context_embeddings, weights

# Calculate for both sentences
# Note: In S1 ["the", "bat", "flew", "over", "his", "head"], 'bat' is at index 1
# Note: In S2 ["he", "hit", "the", "ball", "with", "a", "bat"], 'bat' is at index 6
original_1, context_1, weights_1 = calculate_self_attention(sentence_1, embeddings)
original_2, context_2, weights_2 = calculate_self_attention(sentence_2, embeddings)

# Refined table output

def format_vectors(vectors, decimals=3):
    return [f"[{v[0]:.{decimals}f}, {v[1]:.{decimals}f}]" for v in vectors]


def print_table(sentence, original, context, title):
    df = pd.DataFrame({
        "Word": sentence,
        "Original Vector": format_vectors(original),
        "Contextualized Vector": format_vectors(context)
    })
    print(title)
    print(df.to_string(index=False))


print_table(sentence_1, original_1, context_1, "--- Sentence 1: 'The bat flew over his head' ---")
print()
print_table(sentence_2, original_2, context_2, "--- Sentence 2: 'He hit the ball with a bat' ---")
```

```python
# --- Visualization of the Shift (Sentence-Level, 4 Plots) ---

# Prepare sentence-level embedding arrays
sentence_embeddings_1 = original_1
sentence_context_1 = context_1
sentence_embeddings_2 = original_2
sentence_context_2 = context_2

# Create a 2x2 grid to show original vs. contextualized per sentence
fig, axes = plt.subplots(2, 2, figsize=(14, 10))


def plot_sentence(ax, vectors, sentence, title, extra_points=None):
    for i, word in enumerate(sentence):
        ax.scatter(vectors[i, 0], vectors[i, 1], label=f"{word}")
    if extra_points:
        for point in extra_points:
            ax.scatter(
                point["vector"][0],
                point["vector"][1],
                label=point["label"],
                marker=point.get("marker", "x"),
                color=point.get("color", "black"),
                s=point.get("size", 80)
            )
    ax.set_title(title)
    ax.set_xlabel('Dimension 1')
    ax.set_ylabel('Dimension 2')
    ax.grid(True)
    ax.legend()


# Reference points to emphasize "pulled away from" effect
ball_reference = [{
    "label": "ball (reference)",
    "vector": embeddings["ball"],
    "marker": "x",
    "color": "black",
    "size": 80
}]

flew_reference = [{
    "label": "flew (reference)",
    "vector": embeddings["flew"],
    "marker": "x",
    "color": "black",
    "size": 80
}]

# Sentence 1 plots (include "ball" reference)
plot_sentence(
    axes[0, 0],
    sentence_embeddings_1,
    sentence_1,
    'Original Word Embeddings for "The bat flew over his head"',
    extra_points=ball_reference
)
plot_sentence(
    axes[0, 1],
    sentence_context_1,
    sentence_1,
    'Self-Attention Output Vectors for "The bat flew over his head"',
    extra_points=ball_reference
)

# Sentence 2 plots (include "flew" reference)
plot_sentence(
    axes[1, 0],
    sentence_embeddings_2,
    sentence_2,
    'Original Word Embeddings for "He hit the ball with a bat"',
    extra_points=flew_reference
)
plot_sentence(
    axes[1, 1],
    sentence_context_2,
    sentence_2,
    'Self-Attention Output Vectors for "He hit the ball with a bat"',
    extra_points=flew_reference
)

plt.tight_layout(rect=[0, 0.04, 1, 0.96])
fig.text(
    0.5,
    0.99,
    'Figure 3: Original vs. Self-Attention Output embeddings for both sentences, with reference points to show how "bat" moves away from the irrelevant context word.',
    ha='center',
    fontsize=11
)
plt.show()
```

```python
# --- Validation: Quantitative Check ---
bat_s1 = context_1[1]
bat_s2 = context_2[6]
dist = np.linalg.norm(bat_s1 - bat_s2)
print(f"Euclidean Distance between the two 'bat' contexts: {dist:.4f}")
print("If the distance is > 0, the attention mechanism successfully contextualized the embeddings!")
```

### General Explanation: What do the Vector Values Represent?

**Original Vectors**
*   These are the static, manually initialized representations of the words (e.g., `embeddings = {"bat": [0.5, 0.5]}`).
*   The values `[x, y]` serve as coordinates placing each word in a 2D meaning-space. Words with similar meanings are generally closer together.
*   In the "Original Vector" column of the printed tables, notice that "bat" has the exact same value `[0.500, 0.500]` in both Sentence 1 and Sentence 2. This represents the limitation of static embeddings—they cannot inherently distinguish between the "animal" and the "sports equipment" without context.

**Contextualized Vectors**
*   These are the dynamic outputs of the Self-Attention mechanism. They represent the word *after* it has processed the surrounding context.
*   They are calculated using the attention formula $A = \text{softmax}(QK^T)V$.
    1.  The word "bat" calculates a similarity score (dot product) with every other word in the sentence.
    2.  It assigns high "attention weights" to relevant words (e.g., "flew" in S1 or "ball" in S2).
    3.  The final vector is a weighted sum of the values of those words.
*   **Example**:
    *   In Sentence 1, "bat" (0.5, 0.5) effectively "pulls" the vector of "flew" (0.1, 0.9) into its own representation. The result `[0.203, 0.606]` has shifted heavily towards the Y-axis (high ly), reflecting the "flew" influence.

### Relation to Figure 3 (The 4 Plots)

The values in the matrix of original and contextualized vectors map the geometric positions of the words shown in Figure 3.

*   **Left Column (Original Plots)**: These visuals confirm that the word "bat" occupies the exact same center position $(0.5, 0.5)$ in both scenarios. The model is "confused" or rather, unaware of the specific context.
*   **Right Column (Self-Attention Output Plots)**:
    *   **Top Right (Animal Context)**: The point for "bat" has physically moved up and to the left. It is now clustered near "flew", "head", and "over".
    *   **Bottom Right (Sports Context)**: The point for "bat" has physically moved down and to the right. It is now clustered near "hit" and "ball".
*   **Main Point of the Demonstration**: Figure 3 provides a visual proof that the embeddings are no longer static. The computation has successfully moved the representation of the word into the correct semantic neighborhood.

### Euclidean Distance: Explanation and Interpretation

Euclidean Distance is the straight-line distance between two points in a coordinate system, calculated using the Pythagorean theorem:
$$ d(p, q) = \sqrt{(p_1 - q_1)^2 + (p_2 - q_2)^2} $$

**Calculation for 'Bat':**
Based on the specific results from my results:
*   $\text{Bat}_{context1} \approx [0.203, 0.606]$
*   $\text{Bat}_{context2} \approx [0.517, 0.184]$

$$ \text{Distance} = \sqrt{(0.517 - 0.203)^2 + (0.184 - 0.606)^2} $$
$$ \text{Distance} = \sqrt{(0.314)^2 + (-0.422)^2} $$
$$ \text{Distance} = \sqrt{0.098 + 0.178} \approx \sqrt{0.276} \approx \mathbf{0.526} $$

**Interpretation**:
The calculated distance (approx 0.526) is significantly greater than 0.
*   If the distance were 0, it would mean the context had no effect.
*   Because the distance is > 0, it proves that the model creates a unique numeric identity for "bat" depending on its use. The magnitude (0.526) indicates a strong differentiation between the two meanings in this small vector space.
